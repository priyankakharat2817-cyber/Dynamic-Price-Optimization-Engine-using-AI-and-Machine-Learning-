```python
import zipfile
import os
import pandas as pd

# Unzip the archive.zip file
zip_file_path = '/content/archive.zip'
unzip_dir = '/content/unzipped_data'

if not os.path.exists(unzip_dir):
    os.makedirs(unzip_dir)

with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(unzip_dir)

print(f"Files unzipped to: {unzip_dir}")
print("Contents of unzipped directory:")
for root, dirs, files in os.walk(unzip_dir):
    for file in files:
        print(os.path.join(root, file))

# Assuming the dataset is a CSV file and looking for the first one found
data_file = None
for root, dirs, files in os.walk(unzip_dir):
    for file in files:
        if file.endswith('.csv'):
            data_file = os.path.join(root, file)
            break
    if data_file:
        break

if data_file:
    print(f"\nLoading data from: {data_file}")
    df = pd.read_csv(data_file)
    print("\nFirst 5 rows of the dataset:")
    print(df.head())
    print("\nDataset Info (Before Cleaning):")
    print(df.info())
    print("\nDescriptive Statistics:")
    print(df.describe())

    # Cleaning Step 1: Handle missing values in 'parental_education'
    # Impute missing values in 'parental_education' with the mode
    mode_parental_education = df['parental_education'].mode()[0]
    df['parental_education'] = df['parental_education'].fillna(mode_parental_education)
    print(f"\nMissing values in 'parental_education' imputed with mode: {mode_parental_education}")

    # Cleaning Step 2: Convert categorical columns to 'category' dtype
    categorical_cols = ['gender', 'parental_education', 'internet_access',
                        'extracurricular_activities', 'part_time_job', 'final_grade']
    for col in categorical_cols:
        df[col] = df[col].astype('category')
    print("\nConverted categorical columns to 'category' dtype.")

    print("\nDataset Info (After Cleaning 'parental_education' and converting dtypes):")
    print(df.info())

else:
    print("\nNo CSV file found in the unzipped archive. Please specify the data file to load if it's not a CSV or located elsewhere.")
```
